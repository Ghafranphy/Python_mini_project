```python
import numpy as np
import matplotlib.pyplot as plt

# Generate x values
x = np.linspace(0, 2*np.pi, 100)

# Calculate sine values
y = np.sin(x)

# Plot sine curve
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='sin(x)', linewidth=2)

# Add labels and title
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.title('Sine Curve')

# Add grid and legend
plt.grid(True)
plt.legend()

# Display plot
plt.show()
```
